```python
# YOUR CODE HERE
# Implement the intermediate exercise

class TokenTracker:
    """
    Advanced token tracking with multiple estimation methods.
    """

    def __init__(self, method='word'):
        """
        Initialize tracker with chosen estimation method.

        Args:
            method: 'word' (default) or 'character' estimation
        """
        # Validate the method parameter
        if method not in ['word', 'character']:
            raise ValueError(f"Method must be 'word' or 'character', got '{method}'")

        self.data = []
        self.total_tokens = 0
        self.method = method

    def process(self, input_data):
        """
        Process text and estimate tokens based on selected method.
        """
        # Step 1: Validate input is a string
        if input_data is None:
            return {'status': 'error', 'message': 'Input cannot be None'}

        if not isinstance(input_data, str):
            return {'status': 'error', 'message': f'Input must be a string, got {type(input_data).__name__}'}

        # Step 2: Strip and check if empty
        input_data = input_data.strip()
        if len(input_data) == 0:
            return {'status': 'error', 'message': 'Input cannot be empty'}

        # Step 3: Calculate tokens based on method
        if self.method == 'word':
            # Count words by splitting on whitespace
            token_count = len(input_data.split())
        else:  # method == 'character'
            # Use 4-character rule: characters ÷ 4
            token_count = len(input_data) // 4

        # Step 4: Store the data and update total
        self.data.append({
            'text': input_data,
            'tokens': token_count,
            'method': self.method
        })
        self.total_tokens += token_count

        # Step 5: Return success response
        return {
            'status': 'success',
            'token_count': token_count,
            'total_tokens': self.total_tokens,
            'method': self.method
        }

    def get_entries(self):
        """
        Return all stored entries.
        """
        if len(self.data) == 0:
            return {'status': 'info', 'message': 'No entries stored yet', 'entries': []}

        return {
            'status': 'success',
            'count': len(self.data),
            'entries': self.data
        }

    def get_stats(self):
        """
        Return statistics: total entries, total tokens, average tokens per entry.
        """
        if len(self.data) == 0:
            return {
                'status': 'info',
                'message': 'No data to analyze',
                'total_entries': 0,
                'total_tokens': 0,
                'average_tokens': 0
            }

        avg_tokens = self.total_tokens / len(self.data)

        return {
            'status': 'success',
            'total_entries': len(self.data),
            'total_tokens': self.total_tokens,
            'average_tokens_per_entry': round(avg_tokens, 2),
            'method': self.method
        }


# Test Case 1: Word-based estimation
print("=== WORD METHOD ===")
tracker_word = TokenTracker(method='word')
print(tracker_word.process("Hello world"))
print(tracker_word.process("Python programming is awesome"))
print(tracker_word.get_entries())
print(tracker_word.get_stats())

print("\n=== CHARACTER METHOD ===")
# Test Case 2: Character-based estimation
tracker_char = TokenTracker(method='character')
print(tracker_char.process("Hello world"))  # 11 chars ÷ 4 = 2 tokens
print(tracker_char.process("Python"))  # 6 chars ÷ 4 = 1 token
print(tracker_char.get_entries())
print(tracker_char.get_stats())

print("\n=== ERROR HANDLING ===")
# Test Case 3: Error handling
tracker_test = TokenTracker()
print(tracker_test.process(None))
print(tracker_test.process(123))
print(tracker_test.process(""))
print(tracker_test.get_entries())

# Test Case 4: Invalid method
try:
    bad_tracker = TokenTracker(method='invalid')
except ValueError as e:
    print(f"Caught error: {e}")


pass  # Replace with your implementation
```
